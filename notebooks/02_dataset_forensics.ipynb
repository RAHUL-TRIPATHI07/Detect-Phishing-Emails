```python
import hashlib


archives = {
    "easy_ham": raw_dir / "20030228_easy_ham.tar.bz2",
    "hard_ham": raw_dir / "20030228_hard_ham.tar.bz2",
    "spam": raw_dir / "20030228_spam.tar.bz2",
}

parser = BytesParser(policy=policy.default)

records = []

for label_name, archive_path in archives.items():

    label = 1 if label_name == "spam" else 0

    with tarfile.open(archive_path, "r:bz2") as tar:

        email_files = [
            m for m in tar.getmembers()
            if m.isfile()
        ]

        for i, member in enumerate(email_files):

            raw_bytes = tar.extractfile(member).read()

            msg = parser.parsebytes(raw_bytes)

            subject = msg.get("Subject", "")

            try:
                body_part = msg.get_body(
                    preferencelist=("plain", "html")
                )

                if body_part is not None:
                    body = body_part.get_content()
                else:
                    body = ""

            except Exception:
                body = ""

            subject = str(subject) if subject else ""
            body = str(body) if body else ""

            raw_hash = hashlib.sha256(raw_bytes).hexdigest()

            text = (subject + "\n" + body).strip()

            text_hash = hashlib.sha256(
                text.encode("utf-8", errors="replace")
            ).hexdigest()

            records.append({
                "dataset": label_name,
                "label": label,
                "filename": Path(member.name).name,
                "subject": subject,
                "body": body,
                "text": text,
                "raw_hash": raw_hash,
                "text_hash": text_hash,
            })

            if (i + 1) % 500 == 0:
                print(
                    f"{label_name}: "
                    f"{i + 1}/{len(email_files)} processed"
                )

print("\n" + "=" * 70)
print("Parsing complete")
print("=" * 70)

print("Total records:", len(records))
```

```python
from email import policy
from email.parser import BytesParser
from bs4 import BeautifulSoup

parser = BytesParser(policy=policy.default)


def html_to_text(html):
    """Convert HTML content into readable text."""
    
    soup = BeautifulSoup(html, "html.parser")
    
    # Remove elements that don't represent useful message text
    for tag in soup(["script", "style"]):
        tag.decompose()
    
    return soup.get_text(
        separator=" ",
        strip=True
    )


def extract_email_text(raw_bytes):
    """
    Extract Subject + meaningful message body
    from a raw email.
    """

    msg = parser.parsebytes(raw_bytes)

    # -------------------------
    # Subject
    # -------------------------
    subject = msg.get("Subject", "")
    
    if subject is None:
        subject = ""
    
    subject = str(subject).strip()

    # -------------------------
    # Body
    # -------------------------
    plain_parts = []
    html_parts = []

    if msg.is_multipart():

        for part in msg.walk():

            # Ignore container multipart sections
            if part.is_multipart():
                continue

            # Ignore attachments
            if part.get_content_disposition() == "attachment":
                continue

            content_type = part.get_content_type()

            try:
                content = part.get_content()
            except Exception:
                continue

            if not isinstance(content, str):
                continue

            if content_type == "text/plain":
                plain_parts.append(content)

            elif content_type == "text/html":
                html_parts.append(
                    html_to_text(content)
                )

    else:

        content_type = msg.get_content_type()

        try:
            content = msg.get_content()
        except Exception:
            content = ""

        if isinstance(content, str):

            if content_type == "text/plain":
                plain_parts.append(content)

            elif content_type == "text/html":
                html_parts.append(
                    html_to_text(content)
                )

    # Prefer plain text
    if plain_parts:
        body = "\n".join(plain_parts)

    # Fall back to HTML
    else:
        body = "\n".join(html_parts)

    # -------------------------
    # Combine Subject + Body
    # -------------------------
    text = (
        subject + "\n" + body
    ).strip()

    return text
```

```python
import tarfile
import hashlib
import pandas as pd
from pathlib import Path
from email import policy
from email.parser import BytesParser
from bs4 import BeautifulSoup

raw_dir = Path(r"C:\Users\Rahul\Desktop\Codes\aiml\Machine Learning\Classification\phishingEmail\data\raw")

archives = {
    "easy_ham": raw_dir / "20030228_easy_ham.tar.bz2",
    "hard_ham": raw_dir / "20030228_hard_ham.tar.bz2",
    "spam": raw_dir / "20030228_spam.tar.bz2",
}

parser = BytesParser(policy=policy.default)


def html_to_text(html):
    soup = BeautifulSoup(html, "html.parser")

    for tag in soup(["script", "style"]):
        tag.decompose()

    return soup.get_text(
        separator=" ",
        strip=True
    )


def extract_email_text(raw_bytes):

    msg = parser.parsebytes(raw_bytes)

    # Subject
    subject = msg.get("Subject", "")

    if subject is None:
        subject = ""

    subject = str(subject).strip()

    # Body
    plain_parts = []
    html_parts = []

    if msg.is_multipart():

        for part in msg.walk():

            if part.is_multipart():
                continue

            if part.get_content_disposition() == "attachment":
                continue

            content_type = part.get_content_type()

            try:
                content = part.get_content()
            except Exception:
                continue

            if not isinstance(content, str):
                continue

            if content_type == "text/plain":
                plain_parts.append(content)

            elif content_type == "text/html":
                html_parts.append(
                    html_to_text(content)
                )

    else:

        content_type = msg.get_content_type()

        try:
            content = msg.get_content()
        except Exception:
            content = ""

        if isinstance(content, str):

            if content_type == "text/plain":
                plain_parts.append(content)

            elif content_type == "text/html":
                html_parts.append(
                    html_to_text(content)
                )

    # Prefer plain text
    if plain_parts:
        body = "\n".join(plain_parts)

    else:
        body = "\n".join(html_parts)

    # Subject + body
    text = (
        subject + "\n" + body
    ).strip()

    return subject, body, text


# --------------------------------------------------
# Process complete corpus
# --------------------------------------------------

records = []

for dataset_name, archive_path in archives.items():

    label = 1 if dataset_name == "spam" else 0

    print(f"\nProcessing {dataset_name}...")

    with tarfile.open(archive_path, "r:bz2") as tar:

        email_files = [
            m for m in tar.getmembers()
            if m.isfile()
        ]

        for i, member in enumerate(email_files):

            raw_bytes = tar.extractfile(member).read()

            subject, body, text = extract_email_text(
                raw_bytes
            )

            group_hash = hashlib.sha256(
                text.encode(
                    "utf-8",
                    errors="replace"
                )
            ).hexdigest()

            records.append({
                "dataset": dataset_name,
                "label": label,
                "filename": Path(member.name).name,
                "subject": subject,
                "body": body,
                "text": text,
                "group_hash": group_hash,
            })

            if (i + 1) % 500 == 0:
                print(
                    f"  {i + 1}/{len(email_files)} processed"
                )


df = pd.DataFrame(records)

print("\n" + "=" * 70)
print("FINAL PARSED DATASET")
print("=" * 70)

print("Total rows:", len(df))
print("Columns:", list(df.columns))

print("\nLabel distribution:")
print(df["label"].value_counts().sort_index())

print("\nEmpty extracted text:",
      (df["text"].str.strip() == "").sum())

print("\nUnique groups:",
      df["group_hash"].nunique())
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

# Load frozen splits
train_df = pd.read_parquet(
    r"C:\Users\Rahul\Desktop\Codes\aiml\Machine Learning\Classification\phishingEmail\data\processed\spam_train.parquet"
)

validation_df = pd.read_parquet(
    r"C:\Users\Rahul\Desktop\Codes\aiml\Machine Learning\Classification\phishingEmail\data\processed\spam_validation.parquet"
)

test_df = pd.read_parquet(
    r"C:\Users\Rahul\Desktop\Codes\aiml\Machine Learning\Classification\phishingEmail\data\processed\spam_test.parquet"
)

# Combine TRAIN + VALIDATION
train_validation_df = pd.concat(
    [train_df, validation_df],
    ignore_index=True
)

X_train_validation = train_validation_df["text"]
y_train_validation = train_validation_df["label"]

X_test = test_df["text"]
y_test = test_df["label"]

print("Train + Validation:", len(train_validation_df))
print("Test:", len(test_df))

print("\nTrain + Validation class distribution:")
print(y_train_validation.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())
```

```python
def evaluate_three_zone(lower, upper, scores, labels):
    """
    Three-zone spam decision:
        score < lower       -> NOT SPAM
        lower <= score < upper -> MAYBE SPAM
        score >= upper      -> SPAM
    """

    decisions = np.select(
        [
            scores < lower,
            scores < upper
        ],
        [
            "NOT SPAM",
            "MAYBE SPAM"
        ],
        default="SPAM"
    )

    result = pd.DataFrame({
        "score": scores,
        "label": labels,
        "decision": decisions
    })

    summary = (
        result
        .groupby("decision", sort=False)
        .agg(
            emails=("label", "size"),
            actual_spam=("label", "sum"),
            actual_ham=("label", lambda x: (x == 0).sum())
        )
    )

    summary["spam_rate"] = (
        summary["actual_spam"] / summary["emails"]
    )

    return summary
```
